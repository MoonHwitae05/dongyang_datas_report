```python
# 사용자 입력으로 학점 판정하기
score = input("점수를 입력하세요 (0-100): ")

try:
    score = int(score)
    
    if score >= 90 and score <= 100:
        print("학점: A")
    elif score >= 80 and score < 90:
        print("학점: B")
    elif score >= 70 and score < 80:
        print("학점: C")
    elif score >= 60 and score < 70:
        print("학점: D")
    elif score >= 0 and score < 60:
        print("학점: F")
    else:
        print("오류: 0에서 100 사이의 점수를 입력해주세요.")
except ValueError:
    print("오류: 정수가 아닙니다. 올바른 점수를 입력해주세요.")
```

```python
# match를 사용한 학점 판정 (중간 학점 포함)
score = input("점수를 입력하세요 (0-100): ")

try:
    score = int(score)
    
    if score < 0 or score > 100:
        print("오류: 0에서 100 사이의 점수를 입력해주세요.")
    else:
        # 점수를 5단위로 나누어 match 처리
        match score // 5:
            case 20:  # 100
                grade = "A+"
            case 19:  # 95-99
                grade = "A+"
            case 18:  # 90-94
                grade = "A"
            case 17:  # 85-89
                grade = "B+"
            case 16:  # 80-84
                grade = "B"
            case 15:  # 75-79
                grade = "C+"
            case 14:  # 70-74
                grade = "C"
            case 13:  # 65-69
                grade = "D+"
            case 12:  # 60-64
                grade = "D"
            case _:   # 0-59
                grade = "F"
        
        print(f"학점: {grade}")
except ValueError:
    print("오류: 정수가 아닙니다. 올바른 점수를 입력해주세요.")
```

```python
# match를 사용한 학점 판정 (중간 학점 포함)
score = input("점수를 입력하세요 (0-100): ")

try:
    score = int(score)
    
    if score < 0 or score > 100:
        print("오류: 0에서 100 사이의 점수를 입력해주세요.")
    else:
        # 점수를 10단위로 나누어 match 처리
        match score // 10:
            case 10:  # 100
                grade = "A+"
            case 9:  # 90-99
                match score % 10:
                    case 5 | 6 | 7 | 8 | 9:
                        grade = "A+"
                    case _:
                        grade = "A"
            case 8:  # 80-89
                match score % 10:
                    case 5 | 6 | 7 | 8 | 9:
                        grade = "B+"
                    case _:
                        grade = "B"
            case 7:  # 70-79
                match score % 10:
                    case 5 | 6 | 7 | 8 | 9:
                        grade = "C+"
                    case _:
                        grade = "C"
            case 6:  # 60-69
                match score % 10:
                    case 5 | 6 | 7 | 8 | 9:
                        grade = "D+"
                    case _:
                        grade = "D"
            case _:  # 0-59
                grade = "F"
        
        print(f"학점: {grade}")
except ValueError:
    print("오류: 정수가 아닙니다. 올바른 점수를 입력해주세요.")
```

```python
import matplotlib.pyplot as plt

def analyze_point(point):
    """좌표를 분석하여 메시지 출력"""
    match point:
        case [0, 0]:
            return "원점입니다."
        case [x, 0]:
            return f"X축 위의 점입니다. 위치 X = {x}"
        case [0, y]:
            return f"Y축 위의 점입니다. 위치 Y = {y}"
        case [x, y]:
            return f"일반 2차원 좌표: X={x}, Y={y}"
        case [x, y, z]:
            return f"3차원 좌표입니다: X={x}, Y={y}, Z={z}"
        case _:
            return "단순 2차원 좌표가 아닙니다."

def main():
    print("=" * 50)
    print("좌표 입력 및 시각화 프로그램")
    print("=" * 50)

    points = []  # 입력된 좌표들을 저장할 리스트

    while True:
        print("\n좌표를 입력하세요 (종료하려면 'q' 입력)")
        print("입력 형식: x,y (예: 10,20)")

        user_input = input("좌표: ").strip()

        if user_input.lower() == 'q':
            break

        try:
            # 입력값을 쉼표로 분리하여 좌표로 변환
            coords = [float(x.strip()) for x in user_input.split(',')]

            if len(coords) == 2:
                points.append(coords)
                message = analyze_point(coords)
                print(f"✓ {message}")
            else:
                print("⚠ 2차원 좌표(x,y)를 입력해주세요.")

        except ValueError:
            print("⚠ 올바른 숫자 형식으로 입력해주세요. (예: 10,20)")

    # 좌표 시각화
    if points:
        print(f"\n총 {len(points)}개의 좌표를 입력했습니다.")
        print("그래프를 생성합니다...")

        # 좌표를 x, y 리스트로 분리
        x_coords = [p[0] for p in points]
        y_coords = [p[1] for p in points]

        # 그래프 생성
        plt.figure(figsize=(10, 8))
        plt.scatter(x_coords, y_coords, c='blue', s=100, alpha=0.6, edgecolors='black')

        # 각 점에 좌표 라벨 추가
        for i, (x, y) in enumerate(points):
            plt.annotate(f'({x:.1f}, {y:.1f})',
                        xy=(x, y),
                        xytext=(5, 5),
                        textcoords='offset points',
                        fontsize=9,
                        bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.5))

        # 축과 격자 설정
        plt.axhline(0, color='black', linewidth=0.5)
        plt.axvline(0, color='black', linewidth=0.5)
        plt.grid(True, alpha=0.3)

        plt.xlabel('X 좌표', fontsize=12)
        plt.ylabel('Y 좌표', fontsize=12)
        plt.title('입력한 좌표 시각화', fontsize=14, fontweight='bold')

        plt.tight_layout()
        plt.show()
    else:
        print("\n입력된 좌표가 없습니다.")

if __name__ == "__main__":
    main()
```

```python
# 간단한 몬스터 전투 게임
monster_hp = 100

print("=== 몬스터 전투 시작! ===")
print(f"몬스터 체력: {monster_hp}")

while monster_hp > 0:
    action = input("\n1. 공격하기 | 종료하려면 다른 키 입력: ")

    if action == "1":
        damage = monster_hp * 0.05  # 체력의 5% 데미지
        monster_hp -= damage

        print(f"공격! {damage:.1f}의 데미지!")
        print(f"몬스터 남은 체력: {monster_hp:.1f}")

        if monster_hp <= 0:
            monster_hp = 0
            print(f"\n몬스터 체력: {monster_hp}")
            print("🎉 승리! 몬스터를 물리쳤습니다!")
            break
    else:
        print("게임을 종료합니다.")
        break
```
